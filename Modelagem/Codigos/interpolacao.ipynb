```python
import numpy as np
import matplotlib.pyplot as plt


T = np.array([0, 5, 10, 15, 20, 25, 30])  
O2 = np.array([12.9, 11.3, 10.1, 9.03, 8.17, 7.46, 6.85])  


coeffs = np.polyfit(T, O2, 6)
poly = np.poly1d(coeffs)

T_target = 8
O2_estimated = poly(T_target)

T_range = np.linspace(0, 30, 300)
O2_fit = poly(T_range)

plt.figure(figsize=(8, 5))
plt.scatter(T, O2, color='blue', label='Dados observados')
plt.plot(T_range, O2_fit, color='red', label='Ajuste polinomial (grau 2)')
plt.plot(T_target, O2_estimated, 'go', label=f'O₂ estimado: {O2_estimated:.2f} mg/L')   
plt.title('Ajuste polinomial para O₂ dissolvido (c = 10 g/L)')
plt.xlabel('Temperatura (°C)')
plt.ylabel('Oxigênio dissolvido (mg/L)')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()

O2_estimated, coeffs
```

```python
a, b, c, d, e, f, g = coeffs
print(f"Função ajustada: O₂(T) = "
      f"{a:.4e}·T⁶ + {b:.4e}·T⁵ + {c:.4e}·T⁴ + "
      f"{d:.4e}·T³ + {e:.4e}·T² + {f:.4e}·T + {g:.4e}")
```
